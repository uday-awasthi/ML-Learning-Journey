### 3. Opening a csv file from an URL

```python
import requests
from io  import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

### 4. Sep Parameter

Defines the delimiter used to separate values in a file.
By default, it is , (comma), but it can be changed to other characters like \t (tab), ;, etc.

```python
pd.read_csv('movie_titles_metadata.tsv',sep='\t',names=['sno','name','release_year','rating','votes','geners'])
```

### 5. Index_col parameter

Specifies which column should be used as the row index of the DataFrame instead of default numeric index.

```python
pd.read_csv('aug_train.csv.zip',index_col='enrollee_id')
```

### 6. Header parameter

Defines which row to use as column names.
Default is 0 (first row). You can also set None if the file has no header.

```python
pd.read_csv('train.csv',header =1)
```

### 7. use_cols parameter

Allows selecting specific columns to load from a dataset instead of reading the entire file, reducing memory usage.

```python
pd.read_csv('aug_train.csv.zip', usecols=['enrollee_id','gender','education_level'])
```

### 8. Squeeeze parameters

 If the CSV had only one column, pandas would return a Series instead of a DataFrame.


```python
pd.read_csv('aug_train.csv.zip', usecols=['gender'])
```

### 9. Skiprows/nrows Parameter

skiprows: Skips specified number of rows or list of row indices while reading the file.

nrows: Reads only the first n rows of the dataset.

```python
# pd.read_csv('aug_train.csv.zip',skiprows=[0,5])
pd.read_csv('aug_train.csv.zip',nrows=100)
```

### 10. Encoding parameter

Specifies the text encoding of the file (e.g., utf-8, latin1). It helps in correctly reading special characters.

```python
pd.read_csv('zomato.csv.zip',encoding='latin-1')
```

### 11. Skip bad lines

Used to ignore rows that are malformed or have too many/few columns during reading (mainly in read_csv with on_bad_lines='skip' in newer pandas).

```python
pd.read_csv('BX-Books.csv.zip', sep=';', encoding='latin-1', on_bad_lines='skip')
```

### 12. dtypes parameter

Forces specific data types for columns (e.g., int, float, str) to optimize memory and ensure correct data format.

```python
pd.read_csv('aug_train.csv.zip',dtype={'target':int}).info()
```

### 13. Handling dates

Automatically converts specified columns into datetime format for easier time-series analysis.

```python
pd.read_csv('IPL Matches 2008-2020.csv',parse_dates=['date']).info()
```

```python
def rename(name):
    if name == "Royal Challengers Banglore":
        return "RCB"
    else:
        return name
```

```python
rename("Royal Challengers Banglore")
```

### 14. Converters

Allows applying custom functions to transform values in specific columns while reading the file.

```python
pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename})
```

### 15. na_values parameter

Defines additional strings that should be treated as missing values (NaN), e.g., "NA", "?", "missing".

```python
pd.read_csv('aug_train.csv.zip',na_values=['Male'])
```

### 16. Loading a huge dataset in chunks

Reads large files in smaller parts (chunks) instead of loading everything at once.
Each chunk is a DataFrame, useful for memory-efficient processing.

```python
dfs = pd.read_csv('aug_train.csv.zip',chunksize=5000)
```

```python
for chunk in dfs:
    print(chunk.shape)
```
