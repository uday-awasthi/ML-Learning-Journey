### Fetching data from an API

```python
import pandas as pd
import requests
import time
```

```python

response = requests.get("https://api.themoviedb.org/3/movie/top_rated?api_key=cc65371b62db79455923b71e4be9e64f&language=en-US&page=1")
```

```python
## response.json()['results']
df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
```

```python
df = pd.DataFrame()
```

```python
df
```

```python
for i in range(1, 544):

    try:
        url = f"https://api.themoviedb.org/3/movie/top_rated?api_key=cc65371b62db79455923b71e4be9e64f&language=en-US&page={i}"

        response = requests.get(url, timeout=10)

        temp_df = pd.DataFrame(response.json()['results'])[
            ['id','title','overview','release_date',
             'popularity','vote_average','vote_count']
        ]

        df = pd.concat([df, temp_df], ignore_index=True)

        print(f"Page {i} done")

        time.sleep(1)

    except Exception as e:
        print(f"Error on page {i}: {e}")

print(df.head())
```
