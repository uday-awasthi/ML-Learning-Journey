# Titanic Dataset - Exploratory Data Analysis

In this notebook, we explore the Titanic dataset to understand passenger characteristics and factors affecting survival.

## Import libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train = pd.read_csv('titanic_train.csv')
train.head()
```

## Dataset Overview

Let's examine the shape, columns, and basic information about the dataset.

```python
train.shape
```

```python
train.info()
```

```python
train.describe()
```

## Missing Values Analysis

Before building any model, we need to identify columns with missing values.

```python
train.isnull().sum()
```

```python
sns.countplot(x='Survived', data=train)
plt.show()
```

```python
sns.countplot(x='Sex', hue = 'Survived', data = train)
plt.show()
```

```python
sns.countplot(x='Pclass', hue='Survived', data = train)
plt.show()
```

```python
sns.histplot(train['Age'], kde = True)
plt.show()
```

```python
numeric_df = train.select_dtypes(include = np.number)

plt.figure(figsize=(10,8))
sns.heatmap(numeric_df.corr(), annot=True)
plt.show()
```
