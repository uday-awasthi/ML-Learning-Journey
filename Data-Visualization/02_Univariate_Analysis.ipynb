# EDA using Univariate Analysis

Univariate Analysis means analyzing **one variable at a time**.
It helps us understand the distribution, spread, and pattern of each column individually.

```python
## Imports

 ## **pandas** : data loading and manipulation
import pandas as pd

## **seaborn** : statistical plots
import seaborn as sns

## **matplotlib** : base plotting library
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('train.csv')
```

```python
df.head()
```

## 1. Categorical Data
Data that has fixed categories or labels.

Example: Survived (0/1), Sex (male/female), Pclass (1,2,3)

### a. Countplot

Shows the **count of each category** as bars.
Used to see how many times each category appears in the data.
Example: How many passengers survived vs not survived.

```python
sns.countplot(x = df['Pclass'])
plt.show()
## df['Survived'].value_counts().plot(kind='bar')
```

### b. PieChart

Shows **percentage share** of each category in a circle.
Used when you want to see proportion, not just count.
Example: What % of passengers were male vs female.

```python
df['Sex'].value_counts().plot(kind='pie', autopct='%2f')
```

## 2. Numerical Data
Data that has continuous numbers.

Example: Age, Fare, PassengerId

### a. Histogram

Shows **distribution of a numeric column** by dividing data into bins.
Used to see where most values are concentrated.
Example: Most passengers were aged between 20-40.

```python
plt.hist(df['Age'],bins=5)
```

### b. Distplot

Shows **histogram + KDE (smooth curve)** together.
KDE = Kernel Density Estimate — a smooth version of histogram.
Used to understand the shape of data distribution clearly.
Note: Use `sns.histplot(kde=True)` in new seaborn versions.

```python
# ✅ Histogram + KDE curve (distplot jaisa hi)
sns.histplot(df['Age'], kde=True)
plt.show()
```

### c. Boxplot

Shows **5 key statistics** of a numeric column in one plot:
- Minimum
- Q1 (25th percentile)
- Median (50th percentile)
- Q3 (75th percentile)
- Maximum

Also shows **outliers** as dots outside the box.
Used to detect outliers and understand data spread.


```python
sns.boxplot(df['Age'])
```

```python
## Returns the **minimum value** in the Age column.
## Used to quickly check the range and spot any unusual low values.
df['Age'].min()
```

```python
df['Age'].max()
```

```python
df['Age'].mean()
```

### Skew 

Measures **asymmetry** of the distribution.
- skew = 0 → perfectly symmetric (normal)
- skew > 0 → right skewed (tail on right)
- skew < 0 → left skewed (tail on left)

Important in ML — highly skewed data can hurt model performance.

```python
df['Age'].skew()
```
