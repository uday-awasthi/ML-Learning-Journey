## EDA using Bivariate and Multivariate Analysis

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('tips')
```

```python
titanic = pd.read_csv('train.csv')
```

```python
flights = sns.load_dataset('flights')
```

```python
flights.head()
```

```python
iris = sns.load_dataset('iris')
```

```python
iris.head()
```

### 1. ScatterPlot (Numerical - Numerical)

:A scatter plot visualizes the relationship between two numerical variables.

:Line plots connect data points using lines, making them useful for analyzing trends over time or sequential data. They help visualize increases, decreases, and patterns in the dataset.

```python
sns.scatterplot(x='total_bill', y='tip', data = tips, hue=df['sex'],style=df['smoker'],size=df['size'])
```

### 2. Bar Plot(Numerical - Categorical)

:A bar plot compares values across different categories.

:Bar plots represent categorical data using rectangular bars whose heights or lengths correspond to the values being compared. They are useful for comparing frequencies, averages, or totals among groups.

```python
titanic.head()
```

```python
sns.barplot(x = 'Pclass', y = 'Fare', data = titanic , hue = titanic['Sex'])
```

### 3. Box Plot (Numerical - Categorical)

A box plot summarizes the distribution of a numerical variable and highlights outliers.

A box plot displays the median, quartiles, interquartile range (IQR), and potential outliers. It provides a concise overview of data spread, central tendency, and skewness.

```python
sns.boxplot(x = 'Sex', y = 'Age', data = titanic, hue = titanic['Survived'])
```

### 4. Distplot (Numerical - Categorical)

: A dist plot shows the distribution of a numerical variable using a histogram and density curve, helping understand the data's spread and shape.

```python
sns.histplot(titanic[titanic['Survived']==0]['Age'], kde=True,) #hist=False)
sns.histplot(titanic[titanic['Survived']==1]['Age'], kde=True,) ##hist=False)
```

### 5. HeatMap (Categorical - Categorical)

: A heatmap visualizes correlations between variables using color intensity.

: Heatmaps represent values through color gradients, making it easy to identify strong positive, strong negative, or weak relationships among variables in a correlation matrix.

```python
titanic.head()
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
(titanic.groupby('Embarked').mean(numeric_only=True)['Survived']*100)
```

### 6. ClusterMap(Categorical - Categorical)

: A cluster map combines a heatmap with hierarchical clustering to group similar observations and variables, helping identify patterns in the data.

```python
cross_tab = pd.crosstab(titanic['SibSp'], titanic['Survived'])
sns.clustermap(cross_tab, annot=True, cmap='Blues')
```

### 7. Pairplot

: A pair plot visualizes relationships between multiple numerical variables using scatter plots and distribution plots.

```python
iris.head()
```

```python
sns.pairplot(iris,hue='species')
```

### 8. Lineplot(Numerical - Numerical)

: A line plot shows trends and changes in data over a continuous interval.

: Line plots connect data points using lines, making them useful for analyzing trends over time or sequential data. They help visualize increases, decreases, and patterns in the dataset.

```python
flights.head()
```

```python
new = flights.groupby('year').sum(numeric_only=True).reset_index()
```

```python
sns.lineplot(x = 'year', y ='passengers', data = new)
```
