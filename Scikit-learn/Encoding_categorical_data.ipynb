```python
# Import libraries
import numpy as np
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv('customer.csv')
```

```python
# Check data
print(df.head())
```

```python

# Train Test Split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df[['review', 'education']],
    df['purchased'],      # agar column name purchased hai
    test_size=0.2,
    random_state=42
)
```

What it does:

Splits the dataset into training and testing sets.
X contains input features (review, education).
y contains the target variable (purchased).
80% data is used for training.
20% data is used for testing.

Why?

To evaluate the model on unseen data.

```python
# Ordinal Encoding
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(
    categories=[
        ['Poor', 'Average', 'Good'],
        ['School', 'UG', 'PG']
    ]
)
```

Ordinal Encoding converts ordered categorical values into numerical values.

Mappings:

Review:

Poor    → 0
Average → 1
Good    → 2

Education:

School → 0
UG     → 1
PG     → 2

Why?

Machine learning algorithms work with numbers, not text.

```python
X_train = oe.fit_transform(X_train)
X_test = oe.transform(X_test)
```
